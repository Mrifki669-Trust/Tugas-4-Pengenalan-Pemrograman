```python

# Program Kasir Sederhana

# Meminta input dari pengguna
nama = input("Masukkan nama pelanggan: ")
total_awal = float(input("Masukkan total belanja: Rp"))
diskon = float(input("Masukkan persentase diskon (%): "))

# Menghitung diskon dan total pembayaran
nominal_diskon = total_awal * diskon / 100
total_bayar = total_awal - nominal_diskon

# Menampilkan nota pembayaran
print("\n" + "=" * 36)
print("NOTA PEMBAYARAN".center(36))
print("TOKO SEDERHANA".center(36))
print("-" * 36)

print(f"{nama:^36}")
print("-" * 36)

print(f"{'Total Awal':<20} Rp{total_awal:>13,.2f}")
print(f"{f'Diskon ({diskon}%)':<20} Rp{nominal_diskon:>13,.2f}")
print("-" * 36)
print(f"{'Total Bayar':<20} Rp{total_bayar:>13,.2f}")

print("=" * 36)
print("Terima kasih!".center(36))
print("=" * 36)
```

```python

# Program Pemroses Transkrip Nilai Akademik

# Meminta input dalam satu baris
data = input("Masukkan data mahasiswa (Nama, NIM, Tugas, UTS, UAS): ")

# Memisahkan data menggunakan split
nama, nim, nilai_tugas, nilai_uts, nilai_uas = data.split(",")

# Menghapus spasi dan mengubah nilai menjadi float
nama = nama.strip()
nim = nim.strip()
nilai_tugas = float(nilai_tugas.strip())
nilai_uts = float(nilai_uts.strip())
nilai_uas = float(nilai_uas.strip())

# Menghitung nilai akhir berdasarkan bobot
bobot_tugas = 0.20
bobot_uts = 0.35
bobot_uas = 0.45

nilai_akhir = (
    nilai_tugas * bobot_tugas
    + nilai_uts * bobot_uts
    + nilai_uas * bobot_uas
)

# Menampilkan tabel ringkasan
print("\n" + "=" * 42)
print("TRANSKRIP NILAI MAHASISWA".center(42))
print("=" * 42)

print(f"| {'Nama':<36} |")
print(f"| {nama:<36} |")
print(f"| {'NIM':<36} |")
print(f"| {nim:<36} |")

print("-" * 42)
print(f"| {'Komponen':<14} | {'Nilai':>8} | {'Bobot':>10} |")
print("-" * 42)

print(f"| {'Tugas':<14} | {nilai_tugas:>8.1f} | {bobot_tugas:>9.0%} |")
print(f"| {'UTS':<14} | {nilai_uts:>8.1f} | {bobot_uts:>9.0%} |")
print(f"| {'UAS':<14} | {nilai_uas:>8.1f} | {bobot_uas:>9.0%} |")

print("-" * 42)
print(f"| {'Nilai Akhir':<14} | {nilai_akhir:>8.1f} | {'100%':>10} |")
print("=" * 42)
```
